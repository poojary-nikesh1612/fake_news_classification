```python
import pandas as pd
import numpy as np
import spacy
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import classification_report
```

```python
data=pd.read_csv('Fake_Real_Data.csv')
data.head()
```

```python
data['label_num']=data.label.map({'Fake':0,'Real':1})
print(data.head())
```

```python
nlp=spacy.load("en_core_web_lg")
```

```python
data['vector']=data['Text'].apply(lambda x:nlp(x).vector)
print(data.head())
```

```python
X=np.stack(data.vector)
scaler=MinMaxScaler()
X_scaled=scaler.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_scaled,data.label_num,test_size=0.2,random_state=42,stratify=data.label_num)
```

# Naive Bayes Model

```python
model_nb=MultinomialNB()
model_nb.fit(X_train,y_train)
y_pred_nb=model_nb.predict(X_test)
report_nb=classification_report(y_test,y_pred_nb)
print(report_nb)
```

# KNN Model

```python
model_knn=KNeighborsClassifier(n_neighbors=5,metric='euclidean')
model_knn.fit(X_train,y_train)
y_pred_knn=model_knn.predict(X_test)
report_knn=classification_report(y_test,y_pred_knn)
print(report_knn)
```
